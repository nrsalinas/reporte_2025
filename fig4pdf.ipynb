```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
%matplotlib osx
```

```python
# File paths

# Inputs
cob_ori_file = "chapters/coberturas/dat/Ecosistemas_originaless_v7_light.shp"
cob_act_file = "chapters/coberturas/dat/Ecosistemas_actuales_v5_light.shp"

# Outputs
cob_1 = "fig_1.png"
cob_2 = "fig_2.png"
```

```python
# Mapa coberturas originales
cob_ori = gpd.read_file(cob_ori_file)
cob_ori = cob_ori.to_crs(4326)
fig, ax = plt.subplots(figsize=(30, 15))
cob_ori.plot(column="Eco1", legend=True, ax=ax, legend_kwds={'loc': 'upper left'})
plt.savefig(cob_1, dpi=300, bbox_inches='tight', format='png')
```

```python
# Mapa coberturas actuales
cob_act = gpd.read_file(cob_act_file)
cob_act = cob_act.to_crs(4326)
fig, ax = plt.subplots(figsize=(32, 16))
cob_act.plot(column="ecos_sinte", legend=True, ax=ax, legend_kwds={'loc': 'upper left'})
plt.savefig(cob_2, dpi=300, bbox_inches='tight', format='png')
```
