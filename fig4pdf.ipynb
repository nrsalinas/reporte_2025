```python
# File paths

# Inputs

loc_file = "shared/loca/Loca.shp"

cob_ori_file = "chapters/coberturas/dat/Ecosistemas_originaless_v7_light.shp"
cob_act_file = "chapters/coberturas/dat/Ecosistemas_actuales_v5_light.shp"

fam_file = "chapters/flora/dat/Familias_mas_diversas.csv"
fams = pd.read_csv(fam_file)
gen_file = "chapters/flora/dat/Generos_mas_diversos.csv"
gens = pd.read_csv(gen_file)
spp_flora = "chapters/flora/dat/Especies_localidad.csv"

eco_shape_file = "chapters/plantas_urbanos/dat/shape/ecos.shp"

art_file = "chapters/artropofauna/dat/riqueza_localidad.shp"

# Outputs
base = "pdf_version"
cob_1 = f"{base}/1_coberturas/fig_1.png"
cob_2 = f"{base}/1_coberturas/fig_2.png"
con_1 = f"{base}/2_conectividad/fig_1.pdf"
con_2 = f"{base}/2_conectividad/fig_2.pdf"
con_3 = f"{base}/2_conectividad/fig_3.pdf"
con_4 = f"{base}/2_conectividad/fig_4.pdf"
con_5 = f"{base}/2_conectividad/fig_5.pdf"
flo_1 = f"{base}/3_flora/fig_1.pdf"
flo_2 = f"{base}/3_flora/fig_2.pdf"
flo_3 = f"{base}/3_flora/fig_3.pdf"
flo_4 = f"{base}/3_flora/fig_4.pdf"
flo_5 = f"{base}/3_flora/fig_5.png"

eco_1 = f"{base}/4_ecosistemas_urbanos/fig_1.pdf"
eco_2 = f"{base}/4_ecosistemas_urbanos/fig_2.pdf"
eco_3 = f"{base}/4_ecosistemas_urbanos/fig_3.pdf"
eco_4 = f"{base}/4_ecosistemas_urbanos/fig_4.pdf"
eco_5 = f"{base}/4_ecosistemas_urbanos/fig_5.pdf"
eco_6 = f"{base}/4_ecosistemas_urbanos/fig_6.png"

art_1 = f"{base}/6_artropofauna/fig_1.png"
```

```python
# Mapa coberturas originales
cob_ori = gpd.read_file(cob_ori_file)
cob_ori = cob_ori.to_crs(4326)
fig, ax = plt.subplots(figsize=(30, 15))
cob_ori.plot(column="Eco1", legend=True, ax=ax, legend_kwds={'loc': 'upper left'})
plt.savefig(cob_1, dpi=300, bbox_inches='tight', format='png')
```

```python
# Mapa coberturas actuales
cob_act = gpd.read_file(cob_act_file)
cob_act = cob_act.to_crs(4326)
fig, ax = plt.subplots(figsize=(32, 16))
cob_act.plot(column="ecos_sinte", legend=True, ax=ax, legend_kwds={'loc': 'upper left'})
plt.savefig(cob_2, dpi=300, bbox_inches='tight', format='png')
```

```python
# Mapa artropodos
art = gpd.read_file(art_file)
fig, ax = plt.subplots(figsize=(32, 16))
art.plot(
	ax=ax, 
	column="Riqueza", 
	legend=True, 
	cmap='summer'
)
plt.savefig(art_1, dpi=300, bbox_inches='tight', format='png')
```

```python
# Pies conectividad

co_eep = {
	"Connectividad": ["Área conectada", "Área no conectada"], 
	"Área (ha)": [96649.92, 90097.87]
}
co_eep_pd = pd.DataFrame.from_dict(co_eep)

co_eep_parks = {
	"Connectividad": ["Área conectada", "Área no conectada"], 
	"Área (ha)": [98448.79, 90104.73]
}
co_eep_parks_pd = pd.DataFrame.from_dict(co_eep_parks)

fl_eep = {
	"Clase de área": ["Conector", "Área central"], 
	"Porcentaje": [5, 95]
}
fl_eep_pd = pd.DataFrame.from_dict(fl_eep)

fl_parks = {
	"Clase de área": ["Conector", "Área central"], 
	"Porcentaje": [100, 0]
}
fl_parks_pd = pd.DataFrame.from_dict(fl_parks)

fl_eep_parks = {
	"Clase de área": ["Conector", "Área central"], 
	"Porcentaje": [57, 43]
}
fl_eep_parks_pd = pd.DataFrame.from_dict(fl_eep_parks) 

pie0 = px.pie(
	co_eep_pd, 
	values='Área (ha)', 
	names='Connectividad', 
	#title='Conectividad de la estructura ecológica principal',
	color_discrete_sequence=['#92a311', '#d9c663'],
)
pie0.write_image(con_1)

pie1= px.pie(
	co_eep_parks_pd, 
	values='Área (ha)', 
	names='Connectividad', 
	#title='Conectividad de la estructura ecológica principal + parques urbanos',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie1.write_image(con_2)

pie2 = px.pie(
	fl_eep_pd, 
	values='Porcentaje', 
	names='Clase de área', 
	#title='Tipo de área de la estructura ecológica principal',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie2.write_image(con_3)

pie3 = px.pie(
	fl_parks_pd, 
	values='Porcentaje', 
	names='Clase de área', 
	#title='Tipo de área de los parques urbanos',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie3.write_image(con_4)

pie4 = px.pie(
	fl_eep_parks_pd, 
	values='Porcentaje', 
	names='Clase de área', 
	#title='Tipo de área de la estructura ecológica principal + parques urbanos',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie4.write_image(con_5)
```

```python
###  Flora

ori = {'Origen': ['Introducida', 'Nativa'], 'Porcentaje': [32, 68]}
oripd = pd.DataFrame.from_dict(ori)

cat = {'Categoría': ['En Peligro Crítico', 'En Peligro', 'Vulnerable'], 'Porcentaje': [14, 46, 40]}
catpd = pd.DataFrame.from_dict(cat)

pie1 = px.pie(
	oripd, 
	values='Porcentaje', 
	names='Origen', 
	color_discrete_sequence=['#92a311', '#d9c663'],
	)
pie1.write_image(flo_1)

pie2 = px.pie(
	catpd, 
	values='Porcentaje', 
	names='Categoría', 
	color_discrete_sequence=['#36811c','#859f0c','#e0cd72']
	)
pie2.write_image(flo_2)

bar1 = px.bar(
	fams, 
	x="Familia", 
	y="Número de especies",
	color_discrete_sequence=["#548d10"],
	#height=500
)
bar1.write_image(flo_3)

bar2 = px.bar(
	gens, 
	x="Género", 
	y="Número de especies",
	color_discrete_sequence=["#99a613"],
	#height=500
)
bar2.write_image(flo_4)

loc = gpd.read_file(loc_file)
loc = loc.to_crs(4326)
loc["LocNombre"] = loc.LocNombre.str.title()
loc_spp = pd.read_csv(spp_flora)
loc = loc.merge(
	loc_spp, left_on="LocNombre", right_on="Localidad", how="left"
)
fig, ax = plt.subplots(figsize=(32, 16))
loc.plot(
	ax=ax, 
	column="No. especies", 
	legend=True, 
	cmap='summer'
)
plt.savefig(flo_5, dpi=300, bbox_inches='tight', format='png')
```

```python
# Plantas ecosistemas

encl_ori = {
	'Origen': ['Introducidas', 'Nativas'], 
	'Porcentaje': [8.5, 91.5]
}
encl_ori_pd = pd.DataFrame.from_dict(encl_ori)

hum_ori = {
	'Origen': ['Introducidas', 'Nativas'], 
	'Porcentaje': [41.6, 58.4]
}
hum_ori_pd = pd.DataFrame.from_dict(hum_ori)

parq_ori = {
	'Origen': ['Introducidas', 'Nativas'], 
	'Porcentaje': [47, 53]
}
parq_ori_pd = pd.DataFrame.from_dict(parq_ori)

end = {
	'Ecosistema': ['Enclaves secos', 'Humedales', 'Parques urbanos'],
	'Porcentaje': [17, 24, 63]	
}
end_pd = pd.DataFrame.from_dict(end)

iucn = {
	'Ecosistema': ['Enclaves secos', 'Humedales', 'Parques urbanos'],
	'Peligro Crítico': [0, 6, 2],
	'En Peligro': [0, 8, 8],
	'Vulnerable': [2, 7, 6],
	'Casi Amenazado': [2, 11, 21],
	'Preocupación Menor': [17, 105, 108],
}
iucn_pd = pd.DataFrame.from_dict(iucn)
myyy = [i for i in iucn_pd.columns if i != 'Ecosistema']

pie0 = px.pie(
	encl_ori, 
	values='Porcentaje', 
	names='Origen', 
	#title='Origen geográfico de las plantas de enclaves subxerofíticos',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie0.write_image(eco_1)

pie1 = px.pie(
	hum_ori, 
	values='Porcentaje', 
	names='Origen', 
	#title='Origen geográfico de las plantas de humedales',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie1.write_image(eco_2)

pie2 = px.pie(
	parq_ori, 
	values='Porcentaje', 
	names='Origen', 
	#title='Origen geográfico de las plantas de parques urbanos',
	color_discrete_sequence=['#92a311', '#d9c663']
)
pie2.write_image(eco_3)

pie3 = px.pie(
	end, 
	values='Porcentaje', 
	names='Ecosistema', 
	#title='Endemismo',
	color_discrete_sequence=['#36811c','#859f0c','#e0cd72']
)
pie3.write_image(eco_4)

bar0 = px.bar(
	iucn_pd, 
	x='Ecosistema', 
	y = myyy,
	#title='Plantas amenazadas',
	color_discrete_sequence=["#18351f", "#1f7425", "#809d0b", "#b1af2c", "#f3e9ab"]
)
bar0 = bar0.update_layout(height=500)
bar0.write_image(eco_5)
```
